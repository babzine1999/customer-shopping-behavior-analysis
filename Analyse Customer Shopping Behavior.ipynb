```python
# we habe 37 null value in rating so we need to replace this by median or mean so we will use median because not change with outliers values
data['Review Rating'] = data.groupby('Category')['Review Rating'].transform(lambda x : x.fillna(x.median()))
```

```python
# Renaming columns according to snake casing for better readability and documentation

data.columns = data.columns.str.lower()
data.columns = data.columns.str.replace(' ','_')
data = data.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

Annually → 365 jours
Quarterly → 90 jours (≈ 3 mois)
Monthly → 30 jours
Weekly → 7 jours
Bi-Weekly → 14 jours
Fortnightly → 14 jours
Every 3 Months → 90 jours

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
```

```python
data['purchases frequency days'] = data['frequency_of_purchases'].map(frequency_mapping)
```
