## 3. Data cleaning

### 3.1 Detecting missing values

**Why:** `data.info()` above showed that `Review Rating` has fewer non-null entries (3,863) than the other columns (3,900), which is a signal of missing data. We use `isnull().sum()` to confirm this precisely, column by column, before deciding how to handle it.

```python
data.isnull().sum()
```

**What we found:** the `Review Rating` column has **37 missing values** out of 3,900 rows (about 0.9%). Every other column is fully populated.

**Decision:** we fill these missing values with the **median rating, computed per product category** (`groupby('Category')`), rather than dropping the rows or using a single global value. We use the median instead of the mean because it's more robust to outliers, and we compute it per category rather than globally because average ratings can genuinely differ from one product category to another using a single global median would blur that signal.

```python
# we habe 37 null value in rating so we need to replace this by median or mean so we will use median because not change with outliers values
data['Review Rating'] = data.groupby('Category')['Review Rating'].transform(lambda x : x.fillna(x.median()))
```

```python
data.isnull().sum()
```

✅ **Result:** re-running `isnull().sum()` confirms all missing values have been filled — the dataset is now complete (0 missing values across all columns).

### 3.2 Renaming columns

**Why:** the original column names use mixed case, spaces, and parentheses (e.g. `Purchase Amount (USD)`), which makes them error-prone to reference in code (they require bracket access instead of the cleaner dot notation, and are easy to mistype).

**Decision:** we standardize all column names to **snake_case** (lowercase, underscores instead of spaces), and simplify `purchase_amount_(usd)` to `purchase_amount` for readability.

```python
# Renaming columns according to snake casing for better readability and documentation

data.columns = data.columns.str.lower()
data.columns = data.columns.str.replace(' ','_')
data = data.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
data.head(2)
```

## 4. Feature engineering

### 4.1 Creating an age group (`age_group`)

**Why:** raw `age` is a continuous numeric value, which makes it hard to compare customer segments directly (e.g. "do older customers spend more?"). Grouping customers into age bands makes segment-level analysis and visualization much easier.

**Decision:** we use `pd.qcut` to split customers into **4 equal-sized groups (quartiles)** rather than fixed age ranges, so each segment contains roughly the same number of customers regardless of the age distribution's shape. We label them *Young Adult*, *Adult*, *Middle-aged*, and *Senior*.

**Verification:** we display `age` next to `age_group` to confirm the quartile split is coherent (e.g. 19 years old → *Young Adult*, 63 years old → *Senior*). The values line up as expected.

```python
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
data['age_group'] = pd.qcut(data['age'], q=4, labels = labels)
```

```python
data[['age','age_group']].head(10)
```

### 4.2 Converting purchase frequency to days

**Why:** `frequency_of_purchases` is a free-text categorical column (e.g. "Weekly", "Annually"). Text values can't be used in numeric calculations (e.g. sorting by how often someone buys, or correlating frequency with spending), so we first need to know exactly what categories exist before we can map them to numbers.

```python
data['frequency_of_purchases'].unique()
```

```python
data['frequency_of_purchases'].value_counts()
```

**What we found:** there are 7 distinct frequency categories, fairly evenly distributed (from 539 to 584 customers each) no single frequency dominates.

**Decision:** we convert each text frequency into an approximate number of days, so it becomes usable in numeric calculations (e.g. estimating purchase recency). Note that some categories are treated as equivalent (`Bi-Weekly` ≈ `Fortnightly`, `Quarterly` ≈ `Every 3 Months`):

- Annually → 365 days
- Quarterly → 90 days (≈ 3 months)
- Every 3 Months → 90 days (same as Quarterly)
- Monthly → 30 days
- Bi-Weekly → 14 days
- Fortnightly → 14 days (same as Bi-Weekly)
- Weekly → 7 days

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
```

```python
data['purchases frequency days'] = data['frequency_of_purchases'].map(frequency_mapping)
```

```python
data[['frequency_of_purchases','purchases frequency days']].head(10)
```

## 5. Removing redundant columns

**Why:** while reviewing the columns, `discount_applied` and `promo_code_used` look suspiciously similar a customer with a promo code almost always also has a discount applied. Carrying two columns that encode the same information adds no value and increases the risk of inconsistency later, so we verify whether they're actually identical before deciding whether to drop one.

```python
data.head()
```

```python
(data['discount_applied'] == data['promo_code_used']).all()
```

**What we found:** `(data['discount_applied'] == data['promo_code_used']).all()` returns `True` the two columns are identical across all 3,900 rows.

**Decision:** we drop `promo_code_used` and keep only `discount_applied`, since keeping both would be pure redundancy with no analytical benefit.

```python
# so they are similar
data = data.drop('promo_code_used',axis =1)
```

## 6. Exporting the data to a SQL Server database (optional)

**Why:** exporting the cleaned dataset to a SQL Server database allows it to be queried with SQL, shared with other tools, or integrated into a BI pipeline. This section uses `sqlalchemy` and `pyodbc` to connect to a local SQL Server instance and write the data.


```python
!pip install pyodbc sqlalchemy 
```

* `SQLAlchemy` is a Python library that lets Python talk to databases through a unified interface.
* `create_engine` is the function that creates the bridge between Python and SQL Server:

Python (pandas)  ←→  **create_engine**   ←→  SQL Server

`urllib.parse.quote_plus` is used to URL-encode special characters in the connection string, so that the connection is built correctly.

```python
import pyodbc
print(pyodbc.drivers())
```

```python
from sqlalchemy import create_engine
import urllib
import getpass
import pandas as pd

# Connection
# The password is no longer hard-coded: it's requested securely at runtime instead.
db_password = getpass.getpass("SQL Server password: ")

params = urllib.parse.quote_plus(
    "DRIVER={ODBC Driver 17 for SQL Server};"
    "SERVER=CROWN\\SQLEXPRESS;"
    "DATABASE=customer_behavior_bdd;"
    "UID=sa;"
    f"PWD={db_password};"
    "TrustServerCertificate=yes;"
)

engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")
```

```python

# ETAPE 1 — Écrire les données dans SQL Server
data.to_sql("customer", engine, if_exists="replace", index=False)
print("✅ Table créée avec succès !")

```

```python
# ETAPE 2 — Lire les données
df_result = pd.read_sql("SELECT  * FROM customer;", engine)
print(df_result)
```

```python
pd.read_sql("SELECT TOP 5 * FROM customer;", engine)
```
